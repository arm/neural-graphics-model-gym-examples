# NSS Evaluation Example
---
This notebook shows how to use the Neural Graphics Model Gym to run evaluation of the Neural Super Sampling model.

### Environment Setup
**Before running the notebook:**

Check the environment prerequisites and follow the set up instructions in the [README.md](../../README.md)

```python
# Import the ng_model_gym package.
import ng_model_gym as ngmg
```

### Inspect Configuration Parameters

```python
import logging
from pathlib import Path

# Create a Config object using a path to a config file and extract parameters from it
cfg_path = Path("./assets/tutorial_config.json")

# Load config object from the .json file
config = ngmg.load_config_file(cfg_path)

# Enable minimal logging for ng_model_gym.
# For more detailed logs set log_level to logging.INFO or logging.DEBUG.
ngmg.logging_config(config, log_level=logging.ERROR)
```

```python
from rich import print as rprint

# Let's inspect the loaded config object.
rprint(config)
```

For more information on each of the parameters in the config object, run the following command and note their *description* fields, expected type etc:

```python
ngmg.print_config_options()
```

The config object is mutable, so the default parameters may be overwritten as desired:

```python
# Supply the path to the dataset to run the evaluation on.
config.dataset.path.test = Path("./data/datasets/test/")

config.output.export_frame_png = True
config.model.recurrent_samples = 2
config.dataset.num_workers = 0
```

### Run Evaluation

We can now call the do_evaluate function to evaluate the model using different quality mode presets.

```python
from ng_model_gym import TrainEvalMode

config.model.quality = "high"
config.output.dir = Path("./output")

# Evaluate the Model
ngmg.do_evaluate(
    config, Path("./data/weights/nss_v1_high_int8.pt"), TrainEvalMode.QAT_INT8
)
```

```python
config.model.quality = "mid"
config.output.dir = Path("./output/eval_mid")

# Evaluate the Model
ngmg.do_evaluate(
    config, Path("./data/weights/nss_v1_mid_low_int8.pt"), TrainEvalMode.QAT_INT8
)
```

```python
config.model.quality = "low"
config.output.dir = Path("./output/eval_low")

# Evaluate the Model
ngmg.do_evaluate(
    config, Path("./data/weights/nss_v1_mid_low_int8.pt"), TrainEvalMode.QAT_INT8
)
```

### Visualisation

The ground truth image and the output image from the NSS evaluation will be displayed here. The ground truth image shows the frame fully rendered at native resolution. It represents the best case scenario.

```python
from IPython.display import display, Image, Markdown

FRAME = "0099"

display(Markdown("### Ground Truth"))
display(Image(f"./output/png/ground_truth/frame_{FRAME}_gt.png"))
```

The output image should be close to the ground truth above but the quality of the result will depend on how extensively the model has been trained. We also compare high, mid, and low quality modes below.

```python
# Display the output frames from the NSS evaluation runs
display(Markdown("### High"))
display(Image(f"./output/png/predicted/frame_{FRAME}_pred.png"))

display(Markdown("### Mid"))
display(Image(f"./output/eval_mid/png/predicted/frame_{FRAME}_pred.png"))

display(Markdown("### Low"))
display(Image(f"./output/eval_low/png/predicted/frame_{FRAME}_pred.png"))
```
